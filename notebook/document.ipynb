```python
from langchain_community.document_loaders import SQLDatabaseLoader
from langchain_community.utilities import SQLDatabase
```

```python
import  urllib.parse
user = "root"
password = urllib.parse.quote_plus("Test@123")
host = "127.0.0.1"
port = "3306" # Default MySQL port
database_name = "school"

# 2. Construct the MySQL Connection URL
mysql_uri = f"mysql+pymysql://{user}:{password}@{host}:{port}/{database_name}"

# 3. Initialize the Langchain Database connection
db = SQLDatabase.from_uri(mysql_uri)

# 4. Define the query you want to load data from
query = "SELECT * from students;"

# 5. Initialize the SQLDatabaseLoader
loader = SQLDatabaseLoader(query=query, db=db)

# 6. Load the rows into Langchain Documents
documents = loader.load()

# Test the output
for doc in documents:
    print("Content:", doc.page_content)
    print("Metadata:", doc.metadata)
    print("---")
```
